```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('diabetes.csv')
df.info()
```

```python
print("Nulls")
print("=====")
print(df.isnull().sum())
```

```python
print("0s")
print("==")
print(df.eq(0).sum())
```

```python
columns = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI',
    'DiabetesPedigreeFunction',
    'Age'
]

df[columns] = df[columns].replace(0, np.nan)
```

```python
df.fillna(df.mean(), inplace=True)
```

```python
print(df.eq(0).sum())
```

```python
corr = df.corr()
print(corr)
```

```python
print(df.corr().nlargest(4, 'Outcome').index)
```

```python
print(df.corr().nlargest(4, 'Outcome').values[:,8])
```

```python
%matplotlib inline

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 10))

cax = ax.matshow(corr, cmap='coolwarm', vmin=-1, vmax=1)

fig.colorbar(cax)

ticks = np.arange(0, len(df.columns), 1)

ax.set_xticks(ticks)
ax.set_xticklabels(df.columns)

plt.xticks(rotation=90)

ax.set_yticklabels(df.columns)
ax.set_yticks(ticks)

for i in range(df.shape[1]):
    for j in range(9):
        text = ax.text(
            j, i,
            round(corr.iloc[i][j], 2),
            ha="center",
            va="center",
            color="w"
        )

plt.show()
```

```python
import seaborn as sns

sns.heatmap(df.corr(), annot=True)

fig = plt.gcf()
fig.set_size_inches(8, 8)
```

```python
X = df[['Glucose', 'BMI', 'Age']]
y = df.iloc[:, 8]
```

```python
from sklearn import linear_model
from sklearn.model_selection import cross_val_score

log_regress = linear_model.LogisticRegression()

log_regress_score = cross_val_score(
    log_regress,
    X,
    y,
    cv=10,
    scoring='accuracy'
).mean()

print(log_regress_score)
```

```python
result = []
result.append(log_regress_score)

print(result)
```

```python
from sklearn.neighbors import KNeighborsClassifier

cv_scores = []
folds = 10

ks = list(
    range(
        1,
        int(len(X) * ((folds - 1) / folds)),
        2
    )
)

for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)

    score = cross_val_score(
        knn,
        X,
        y,
        cv=folds,
        scoring='accuracy'
    ).mean()

    cv_scores.append(score)

knn_score = max(cv_scores)

optimal_k = ks[cv_scores.index(knn_score)]

print(f"The optimal number of neighbors is {optimal_k}")
print(knn_score)

result.append(knn_score)
```

```python
from sklearn import svm

linear_svm = svm.SVC(kernel='linear')

linear_svm_score = cross_val_score(
    linear_svm,
    X,
    y,
    cv=10,
    scoring='accuracy'
).mean()

print(linear_svm_score)

result.append(linear_svm_score)
```

```python
rbf = svm.SVC(kernel='rbf')

rbf_score = cross_val_score(
    rbf,
    X,
    y,
    cv=10,
    scoring='accuracy'
).mean()

print(rbf_score)

result.append(rbf_score)
```

```python
algorithms = [
    "Logistic Regression",
    "K Nearest Neighbors",
    "SVM Linear Kernel",
    "SVM RBF Kernel"
]

cv_mean = pd.DataFrame(result, index=algorithms)

cv_mean.columns = ["Accuracy"]

cv_mean.sort_values(
    by="Accuracy",
    ascending=False
)
```

```python
knn = KNeighborsClassifier(n_neighbors=19)

knn.fit(X, y)
```

```python
import pickle

filename = 'diabetes.sav'

pickle.dump(knn, open(filename, 'wb'))
```

```python
loaded_model = pickle.load(open(filename, 'rb'))
Glucose = 65
BMI = 70
Age = 50

prediction = loaded_model.predict([[Glucose, BMI, Age]])

print(prediction)

if prediction[0] == 0:
    print("Non-diabetic")
else:
    print("Diabetic")
```

```python
proba = loaded_model.predict_proba([[Glucose, BMI, Age]])

print(proba)

print(
    "Confidence: " +
    str(round(np.amax(proba[0]) * 100, 2)) +
    "%"
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.metrics import confusion_matrix, roc_curve, auc
from sklearn.model_selection import learning_curve
```

```python
# Cấu hình phong cách đồ thị học thuật
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams.update({'font.size': 11, 'figure.autolayout': True})
```

```python
# 1. Biểu đồ phân phối biến mục tiêu (Target Distribution)
plt.figure(figsize=(6, 4))
ax = sns.countplot(x=df['Outcome'], palette='Set2')
plt.title('Phân bố biến mục tiêu (Outcome - Bệnh tiểu đường)', fontsize=13, fontweight='bold')
plt.xlabel('Trạng thái (0: Không mắc bệnh, 1: Mắc bệnh)')
plt.ylabel('Số lượng mẫu')
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='baseline', fontsize=11, color='black', xytext=(0, 3), textcoords='offset points')
plt.savefig('diabetes_target_dist.png', dpi=300)
plt.show()
```

```python
# 2. Biểu đồ Boxplot phân tích ngoại lệ (Outlier Analysis) theo biến mục tiêu
features_to_box = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'Age']
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
for i, col in enumerate(features_to_box):
    sns.boxplot(x='Outcome', y=col, data=df, ax=axes[i//3, i%3], palette='pastel')
    axes[i//3, i%3].set_title(f'Phân phối {col} theo Outcome', fontweight='bold')
plt.savefig('diabetes_boxplots.png', dpi=300)
plt.show()
```

```python
# 3. Phân phối tần suất các biến số (Histogram + KDE)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for i, col in enumerate(['Glucose', 'BMI', 'Age']):
    sns.histplot(df[col], kde=True, ax=axes[i], color='teal', bins=25)
    axes[i].set_title(f'Phân phối tần suất {col}', fontweight='bold')
plt.savefig('diabetes_histograms.png', dpi=300)
plt.show()
```

```python
# 4. Biểu đồ so sánh hiệu năng các mô hình (Model Comparison Barplot)
plt.figure(figsize=(8, 4))
plot_acc = cv_mean.reset_index()
plot_acc.columns = ['Model', 'Accuracy']
ax = sns.barplot(x='Accuracy', y='Model', data=plot_acc.sort_values('Accuracy', ascending=False), palette='Blues_r')
plt.title('So sánh độ chính xác (Accuracy 10-Fold CV) giữa 4 mô hình', fontsize=13, fontweight='bold')
plt.xlabel('Accuracy trung bình')
for p in ax.patches:
    ax.annotate(f"{p.get_width():.4f}", (p.get_width(), p.get_y() + p.get_height() / 2.),
                ha='left', va='center', xytext=(5, 0), textcoords='offset points', fontweight='bold')
plt.xlim(0.7, 0.82)
plt.savefig('diabetes_model_comparison.png', dpi=300)
plt.show()
```

```python
# 5. Ma trận nhầm lẫn (Confusion Matrix) của mô hình tối ưu (KNN k=19)
from sklearn.model_selection import cross_val_predict
y_pred_cv = cross_val_predict(knn, X, y, cv=10)
cm = confusion_matrix(y, y_pred_cv)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Dự đoán: 0 (Âm tính)', 'Dự đoán: 1 (Dương tính)'],
            yticklabels=['Thực tế: 0', 'Thực tế: 1'])
plt.title('Ma trận nhầm lẫn (Confusion Matrix) - KNN (k=19)', fontsize=13, fontweight='bold')
plt.savefig('diabetes_confusion_matrix.png', dpi=300)
plt.show()
```

```python
# 6. Đường cong ROC và diện tích dưới đường cong (ROC Curve & AUC)
y_proba_cv = cross_val_predict(knn, X, y, cv=10, method='predict_proba')[:, 1]
fpr, tpr, _ = roc_curve(y, y_proba_cv)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'Đường cong ROC (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Ngẫu nhiên (AUC = 0.5000)')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Tỷ lệ dương tính giả (False Positive Rate)')
plt.ylabel('Tỷ lệ dương tính thật (True Positive Rate)')
plt.title('Đường cong ROC-AUC mô hình KNN (10-Fold CV)', fontsize=13, fontweight='bold')
plt.legend(loc='lower right')
plt.savefig('diabetes_roc_curve.png', dpi=300)
plt.show()
```
